```python
# resize frames to fit model 512*512
# original is 1920 * 1080
# resize to 512 * 288 then add padding to 512 * 512

import cv2
import matplotlib.pyplot as plt

new_width = 512
new_height = 288
mask_file = "../data/unet_exp1/masks/S57O_frame01.png"

mask = cv2.imread(mask_file, cv2.IMREAD_GRAYSCALE)
resized_mask = cv2.resize(
    mask,
    (new_width, new_height),
    interpolation=cv2.INTER_NEAREST
)

plt.imshow(resized_mask, cmap="gray")
plt.show()
plt.imshow(mask, cmap="gray")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

new_width = 512
new_height = 288
frame_file = "../data/unet_exp1/frames/S57O_frame01.png"

frame = cv2.imread(frame_file, cv2.IMREAD_UNCHANGED)
resized_frame = cv2.resize(
    frame,
    (new_width, new_height),
    interpolation=cv2.INTER_NEAREST
)

plt.imshow(resized_frame)
plt.show()
plt.imshow(frame, cmap="gray")
plt.show()
```

```python
# add 2 pads top and bottom 112 each

padded_mask = cv2.copyMakeBorder(
    resized_mask,
    top=112,
    bottom=112,
    left=0,
    right=0,
    borderType=cv2.BORDER_CONSTANT,
    value=0        # black padding — use 0 for masks
)

plt.imshow(padded_mask, cmap="gray")
plt.show()
```
