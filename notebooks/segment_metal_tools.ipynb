```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
frame0 = cv2.imread("./frame_result/frame0.jpg")
```

```python
color_conversions = [flag for flag in dir(cv2) if flag.startswith('COLOR_BGR')]
print(color_conversions)
print(len(color_conversions))
```

```python
# try all possible cv2 color converter -- do not look very much helpful
plt.figure(figsize=(90, 60))
x = 1
for flag_name in color_conversions:
    frame0 = cv2.imread("./frame_result/frame0.jpg")
    try:
        converted = cv2.cvtColor(frame0, getattr(cv2, flag_name))
        plt.subplot(9, 6, x)
        plt.imshow(converted)
        plt.title(flag_name, fontsize=7)
    except Exception as e:
        plt.subplot(9, 6, x)
        plt.imshow(frame0, cmap = 'gray')
        plt.title(flag_name, fontsize=7)
    x += 1
```

```python
# Edge dectection with pillow - very obvious for thread and three equipments
# source: https://www.geeksforgeeks.org/python/python-edge-detection-using-pillow/
from PIL import Image, ImageFilter

image = Image.open("./frame_result/frame0.jpg")
image = image.convert("L")
image = image.filter(ImageFilter.FIND_EDGES)

plt.imshow(image, cmap = 'gray')
plt.show()

plt.imshow(image)
plt.show()
```
