# Subtlety of Train and Test Split of this Dataset
NOTE: This is a follow up from **analyze_dataset.ipynb**

Data 314 vidoes: 214 (Train), 100 (Test)
1. PRE & POST Split
   1. PRE 157 vids: 107 (Train), 50 (Test)
   2. POST 157 vids: 107 (Train), 50 (Test)
2. Similar amount of all skill levels
   1. Organize it to Novice, Intermediate, Proficient
   2. Catagorize based on average GRS of 3 evaluaters
      1. WHY: Since alpha value of 3 of them is 0.73

This method is called **stratified sampling**

**Purpose for spliting the data:**
- extract numerical features from training videos
- develop the mathematical model from training videos
- test it on testing videos

#### Next steps:
1. Identify videos for pre
2. Identify videos for post
3. Identify videos for Novice, Intermediate, Proficient

```python
file = "../data/osats_score.csv"
```

```python
import csv

def _read_pre_post(type: str, file: str) -> dict:
    """Return a list of data read form given csv file that has
    type tag ('PRE' or 'POST'). The returned list contains
    list of student_id and average grs.
    EX: {'AHO729': [13, 14, 15]}
    """
    output = {}
    with open(file, "r") as f:
        data = csv.reader(f)
        next(data)
        for item in data:
            if item[2] == type:
                if item[0] not in output:
                    output[item[0]] = [int(item[-1])]
                else:
                    output[item[0]].append(int(item[-1]))
    return output

def _calculate_avg_grs(vid_to_score: dict) -> dict:
    """Return a dict, which key is a student_id and value is avg grs"""
    output = {}
    for item in vid_to_score:
        avg_grs = sum(vid_to_score[item])/3
        output[item] = avg_grs
    return output

def create_student_id_split(type: str, file: str) -> dict:
    """Return a dict that looks like the following:
    {novice: [student_id], intermediate: [student_id], proficient: [student_id]}
    """
    vid_to_score = _read_pre_post(type, file)
    avg_grs_per_vid = _calculate_avg_grs(vid_to_score)
    output = {'novice': [], 'intermediate': [], 'proficient': []}
    for id in avg_grs_per_vid:
        if avg_grs_per_vid[id] >= 24:
            output['proficient'].append(id)
        elif avg_grs_per_vid[id] >= 16:
            output['intermediate'].append(id)
        else:
            output['novice'].append(id)
    return output
```

```python
all_pre = create_student_id_split('PRE', file)
all_post = create_student_id_split('POST', file)

print(f"PRE: {all_pre}")
print(f"POST: {all_post}")
```

### Randomly pick videos based on defined amount

```python
# Where these numbers are from will be written in the manuscript

amount_train_split_pre = {'novice': 78, 'intermediate': 25, 'proficient': 4}
amount_train_split_post = {'novice': 1, 'intermediate': 26, 'proficient': 80}
```

```python
import random

train_pre = {'novice': [], 'intermediate': [], 'proficient': []}
for item in all_pre:
    train_pre[item] = random.sample(all_pre[item], k=amount_train_split_pre[item])

print(train_pre)
print(len(train_pre['novice']))
```

```python
test_pre = {'novice': [], 'intermediate': [], 'proficient': []}
for item in all_pre:
    for id in all_pre[item]:
        if id not in train_pre[item]:
            test_pre[item].append(id)

print(test_pre)
```

```python
import random

def split_train(all_data: dict, amount_split: dict) -> dict:
    """Return a tuple of the split of train and test"""
    train_data = {'novice': [], 'intermediate': [], 'proficient': []}
    for item in all_data:
        train_data[item] = random.sample(all_data[item], k=amount_split[item])
    return train_data

def split_test_based_on_train(all_data: dict, train_data: dict) -> dict:
    """Return a tuple of the split of train and test"""
    test_data = {'novice': [], 'intermediate': [], 'proficient': []}
    for item in all_data:
        for id in all_data[item]:
            if id not in train_data[item]:
                test_data[item].append(id)
    return test_data
```

### Combine everything again

```python
# fixed data
file = "../data/osats_score.csv"
amount_train_split_pre = {'novice': 78, 'intermediate': 25, 'proficient': 4}
amount_train_split_post = {'novice': 1, 'intermediate': 26, 'proficient': 80}

# run based on functions built
all_pre = create_student_id_split('PRE', file)
all_post = create_student_id_split('POST', file)

pre_train = split_train(all_pre, amount_train_split_pre)
pre_test = split_test_based_on_train(all_pre, pre_train)
post_train = split_train(all_post, amount_train_split_post)
post_test = split_test_based_on_train(all_post, post_train)
```
